# padding



#cnn without padding (valid)

```python
import tensorflow
from tensorflow import keras
from keras.layers import Dense,Conv2D,Flatten
from keras import Sequential
from keras.datasets import mnist
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
model = Sequential()

model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu', input_shape=(28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))
#activation relu = to make computaion easier big values = 255 small =0
model.add(Flatten()) #converting dat a to 1D

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary() #reuslts with mxm  *  nxn = m-n+1
```

cnn with padding (same)

```python
model1 = Sequential()

model1.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu', input_shape=(28,28,1)))
model1.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu'))
model1.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu'))

model1.add(Flatten())

model1.add(Dense(128,activation='relu'))
model1.add(Dense(10,activation='softmax'))
```

# strides

cnn without padding and normal stride value (low level feature extarction)

```python
model3 = Sequential() #formula = [(n-f)/s =1]

model3.add(Conv2D(32,kernel_size=(3,3),padding='valid',strides=(1,1), activation='relu', input_shape=(28,28,1)))
model3.add(Conv2D(32,kernel_size=(3,3),padding='valid',strides=(1,1), activation='relu'))
model3.add(Conv2D(32,kernel_size=(3,3),padding='valid',strides=(1,1), activation='relu'))

model3.add(Flatten())

model3.add(Dense(128,activation='relu'))
model3.add(Dense(10,activation='softmax'))
```

```python
model3.summary()
```

cnn with padding and strided covnvolution (skipping pixels high level feature extarction)

```python
model4 = Sequential() #formula = [(n+2p-f)/s =1]

model4.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu', input_shape=(28,28,1)))
model4.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu'))
model4.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu'))

model4.add(Flatten())

model4.add(Dense(128,activation='relu'))
model4.add(Dense(10,activation='softmax'))
```

```python
model4.summary()
```

```python
#high value of strides = info loss
#less value of stride = less loss of info
```
