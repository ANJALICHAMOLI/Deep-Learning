```python
from keras import Sequential
from keras.layers import Dense,SimpleRNN
```

```python
model = Sequential()
model.add(SimpleRNN(3,input_shape=(4,5)))
#simple rnn model with hidden layer of 3 nurons and input dhape (4,5) tensor as 4 is no of time steps (longest sentence len)
#and 5 is inputfeatures(list of one word [1 0 0 0 0])
model.add(Dense(1))
#dense layer with 1 nuron
model.summary()
```

the final trainable parameters analysis

weights = 5x3=15

then output of the hidden layer for time step 2 then 3x3=9 weights

the final layer as in output last nuron

3x1=3 weights

= 27 weights


baise of hiddden layer = 3
and output layer = 1

baises = 4
