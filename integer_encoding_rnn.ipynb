```python
import numpy as np

docs = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'inquilab zindabad']
```

```python
# step1: tokenizing sentencese
#Tokenizer creates a vocabulary and assigns an integer to each word.
# includes-word by word seggregation
#         -cleaning text by converting upper to lower case
#         or removing special symbols or punctuations
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer(oov_token='<oov>')
# ovv= out of vocabulary-->it is a special placeholder used when the tokenizer
# encounters a word during testing/prediction that it did not see during training.
#so it replacese the unseen work with the oov token
```

```python
tokenizer.fit_on_texts(docs)
#segrregates word by word
#assigns indexs etc
```

```python
tokenizer.word_index #index assigned to eveyr unique word
```

```python
tokenizer.word_counts #word freqeuncy
```

```python
tokenizer.document_count #total number of sentencese provided or fed
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences
```

```python
sequences = pad_sequences(sequences,padding='post') #post menaig after sequce padding pre means before sequnce padding
```

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train[0] #already tokenized and sequnced basically integer encoded
```

```python
len(X_train[1])
```

```python
len(X_train[5])
```

```python
#since every vector diff len so we use padding
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
# every reviwe will be trimmed to 50 words due to large reviwe
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
X_train.shape
```

```python
model = Sequential()

model.add(SimpleRNN(32,input_shape=(50,1),return_sequences=False))
#this 50 = timestep
#with every time step one word(no number) will be sent to the model
#retn_sequnce flase so that we dong get output seprately
model.add(Dense(1,activation='sigmoid'))
#CAL OF WEIGHTS AND BAIS FOR EVEYR LAYER
# SimpleRNN (1×32) + (32×32) + 32=1088
#Dense (32x1)+1 = 33

model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])

model.fit(X_train,y_train,epochs=5,validation_data=(X_test,y_test))
```
