```python
import tensorflow as tf
from keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
from google.colab import files
files.upload()
```

```python
img = image.load_img('cat_img.jpg',target_size=(200,200))
```

```python

datagen = ImageDataGenerator(
        rotation_range=30,
        width_shift_range=0.2,
        height_shift_range=0.2,
        rescale=1./255,
        shear_range=0.2,
        zoom_range=0.2,
        horizontal_flip=True,
        fill_mode='nearest')
```

```python
img=image.img_to_array(img)
```

```python
type(img)
img.shape
```

```python
input=img.reshape((1,200,200,3))
```

```python
import shutil

shutil.rmtree('/content/aug')
```

```python
import os
os.makedirs('/content/aug', exist_ok=True)
```

```python
i=0
for output in datagen.flow(input,batch_size=1,save_to_dir='/content/aug'):
  i=i+1
  if i==10:
    break
```
