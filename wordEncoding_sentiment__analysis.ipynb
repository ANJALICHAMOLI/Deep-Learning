```python
#word encoding - embbeding technique used for sentiment analysis
```

```python
docs = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'inquilab zindabad']
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts(docs)
```

```python
len(tokenizer.word_index) #total unique words in vocab
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences
sequences = pad_sequences(sequences,padding='post')
sequences
```

```python
from keras.datasets import imdb
from keras.utils import pad_sequences
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Input,Flatten
```

```python
model = Sequential()
model.add(Input(shape=(5,)))
model.add(Embedding(15,output_dim=2))
#embedding layer added to sequential model with para = (total unique words,output vector of the mebedding(dense vector),no of words)
#output dimis hyper parameter which can be tuned to get best value the dim given is--
# basically each word becomes for eg 2 dim meaning every word will be like "go" → [0.2 , 0.7]
#input_len =each input sentence has 5 words/
# eg "I love AI" after tokenizing [1,2,3]  then padding since input len =5 [0,0,1,2,3]

model.summary()
```

```python
model.compile('adam','accuracy')
```

```python
pred = model.predict(sequences)
print(pred) #representations of review
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
X_train.shape
```

```python
from keras import Sequential
```

```python
model = Sequential()
model.add(Input(shape=(50,)))
model.add(Embedding(input_dim=10000,output_dim=2,input_length=50))
model.add(SimpleRNN(32,return_sequences=False))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])
history = model.fit(X_train, y_train,epochs=5,validation_data=(X_test,y_test))
```
